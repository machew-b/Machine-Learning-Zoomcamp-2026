## Q6. Median value of horsepower

1. Find the median value of `horsepower`.
2. Find the most frequent value (mode) of `horsepower`.
3. Fill missing values in `horsepower` with that mode.
4. Recompute the median. Did it change?

```python
median_before = df['horsepower'].median()
most_frequent = df['horsepower'].mode()[0]
median_before, most_frequent
```

```python
horsepower_filled = df['horsepower'].fillna(most_frequent)
median_after = horsepower_filled.median()
median_after
```

```python
if median_after > median_before:
    print("Yes, it increased")
elif median_after < median_before:
    print("Yes, it decreased")
else:
    print("No")
```

## Q7. Sum of weights

A mini linear regression via the normal equation:
1. Select cars from Asia only.
2. Keep only `vehicle_weight` and `model_year`.
3. Take the first 7 rows.
4. Convert to a NumPy array `X`.
5. Compute `XTX = X.T @ X`, then invert it.
6. Let `y = [1100, 1300, 800, 900, 1000, 1100, 1200]`.
7. Compute `w = XTX_inv @ X.T @ y`.
8. Sum the elements of `w`.

```python
asia = df[df['origin'] == 'Asia']
subset = asia[['vehicle_weight', 'model_year']].head(7)
subset
```

```python
X = subset.values
XTX = X.T @ X
XTX_inv = np.linalg.inv(XTX)

y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
w
```

```python
w.sum()
```
